# Transcriptor de audio

## 1. Importar librerías

```python
import whisper # Modelo de Openai para reconocimiento automático del habla (ASR)
import torch # Motor que soporta y ejecuta el modelo de Whisper de forma eficiente utilizando la GPU
import time # Para medir cuanto tarda el modelo en procesar y transcribir un archivo de audio
import os # Para interactuar con el sistema operativo (gestión de carpetas, archivos, directorios...)
import imageio_ffmpeg # Necesario para que Whisper extraiga y decodifique el flujo de audio de diferentes formatos
import sys
```

## Añadir FFmpeg al PATH de Python

```python

# Añade al PATH del proceso de Python la carpeta Scripts de nuestro entorno, 
# donde se encuentra ffmpeg.exe, para que Whisper pueda encontrar y utilizar FFmpeg al cargar los archivos de audio.

# Obtiene la carpeta donde está instalado el entorno de Python
carpeta_entorno = os.path.dirname(sys.executable)

# Construye la ruta hasta la carpeta Scripts del entorno
carpeta_scripts = os.path.join(carpeta_entorno, "Scripts")

os.environ["PATH"] += os.pathsep + carpeta_scripts

# os.environ["PATH"] → accede al PATH que está utilizando Python
# += → añade información al contenido que ya existe
# os.pathsep → utiliza el separador correcto de rutas de Windows
```

## 2. Cargar el modelo de IA de Whisper en la memoria

```python
modelo = whisper.load_model("medium", device = "cuda")
```

## 3. Crear la ruta de los archivos que se van a transcribir

```python
archivos = os.listdir('audios')

ruta_de_archivos = []

for archivo in archivos:
    ruta = os.path.join('audios', archivo)
    ruta_de_archivos.append(ruta)

ruta_de_archivos
```

##  4. Transcribir cada audio, medir el tiempo que tarda cada uno y guardar cada transcripción en un archivo de texto

```python
transcripciones = {}

os.makedirs("transcripciones", exist_ok = True)

for ruta in ruta_de_archivos:

# Inicia el tiempo de transcripción
    inicio = time.time()

    print(ruta)
    
# Inicia la transcripción
    resultado = modelo.transcribe(
        ruta,
        language="es")
    
# Finaliza el tiempo de transcripción
    fin = time.time()

# Calcula la duración de la transcripción
    print(f"Tiempo: {fin-inicio: .2f} segundos.")

# Guarda en el diccionario la ruta como clave y el texto de la transcripción como valor
    transcripciones[ruta] = resultado['text']

# Elimina del nombre de la ruta la carpeta donde se guarda el audio
    nombre_archivo = os.path.basename(ruta)

# Divide el nombre del audio en nombre del archivo y su extensión (.m4a)
    nombre, extension = os.path.splitext(nombre_archivo)

# Al nombre del archivo se le añade .txt para crear el archivo de texto de la transcripción
    archivo_txt = nombre + ".txt"

# Crea la ruta en la que se van a guardar los archivos.txt
    ruta_txt = os.path.join('transcripciones', archivo_txt)

# Guarda la transcripción en un archivo de texto
    with open(ruta_txt, "w", encoding="utf-8") as archivo:
        archivo.write(resultado["text"])
```
